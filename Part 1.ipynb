```python
#stap 0: libraries importeren en dataset laden
import pandas as pd # dataset inladen en bewerken
import numpy as np # numeriek berekeningen
import matplotlib.pyplot as plt # grafieken maken
import seaborn as sns # statistiche visualisatie maken

## load dataset 
df = pd.read_csv("NHANES_20212023_merged_filtered 1.csv",
                 encoding="cp1252")
```

```python
#stap 1: Algemene exploratie van de dataset

## hoeveel samples en features 
print("Number of samples:", df.shape[0]) # shape= geef afmetingen van een datastructuur terug in de vorm van een tuple
print("Number of features:", df.shape[1])
```

```python
## bekijk eerste rijen 
df.head()
```

```python
## bekijk alle kolomnamen 
df.columns.tolist()
```

```python
## bekijk algemene structuur 
df.info()
```

```python
# datatype (in totaal 3 verschillende)
df.dtypes.value_counts()
```

```python
#stap 2: feature summary maken
feature_summary = pd.DataFrame({
    "dtype": df.dtypes,
    "n_unique": df.nunique(), # aantal verschillende mogelijkheden niet missing waarden
    "missing_count": df.isnull().sum(), # aantal ontbrekende waarden
    "missing_percentage": df.isnull().mean() * 100
})

display(feature_summary)
```

```python
## bereken percentage missing values
missing_percentage = df.isnull().mean() * 100

missing_percentage.sort_values(ascending=False).head(20)
```

```python
#stap 3: Missing values onderzoeken

missing_count = df.isnull().sum()
## 20 features met de meeste missing values
missing_count.sort_values(ascending=False).head(20)
```

```python
## 20 features met de minst missing values
missing_count.sort_values().head(20)
```

```python
#stap 4: fouten in dataset opsporen 
## duplicate controlen
print("Duplicate rows:", df.duplicated().sum())

# Check duplicate participant IDs
print("Duplicate participant IDs:", df["SEQN"].duplicated().sum())
```

```python
## constante features onderzoeken = heeft voor alle deelnemers dezelfde geldige waarde 
constant_features = feature_summary[
    feature_summary["n_unique"] <= 1
]

display(constant_features)
```

```python
## features met bijna alleen missing values 
high_missing_features = feature_summary[
    feature_summary["missing_percentage"] >= 95
]

display(high_missing_features)
```
